```python
# Generate simulated temperature readings from two sensors.
# Sensor A: mean 25 C, std 3 C, 200 readings.
# Sensor B: mean 27 C, std 4.5 C, 200 readings.
# Also generate 200 timestamps uniformly from 0 to 10 seconds.
# Use np.random.default_rng with a seed = last 4 digits of your Drexel ID

import numpy as np
import matplotlib.pyplot as plt

def main():
    # Parameters
    n = 200
    seed = 1234 # REPLACE 1234 with the last 4 digits of your Drexel ID
    rng = np.random.default_rng(seed)

    # Data generation per intent comment
    t = np.sort(rng.uniform(0.0, 10.0, n))          # 200 timestamps uniformly 0–10 s, sorted
    sensor_A = rng.normal(loc=25.0, scale=3.0, size=n)
    sensor_B = rng.normal(loc=27.0, scale=4.5, size=n)

    # Basic plot
    plt.figure(figsize=(9, 4))
    plt.plot(t, sensor_A, 'o', ms=4, alpha=0.75, label='Sensor A (μ=25, σ=3)')
    plt.plot(t, sensor_B, 's', ms=4, alpha=0.75, label='Sensor B (μ=27, σ=4.5)')
    plt.axhline(25, color='C0', ls='--', lw=1)
    plt.axhline(27, color='C1', ls='--', lw=1)
    plt.xlabel('Time (s)')
    plt.ylabel('Temperature (°C)')
    plt.title('Simulated temperature readings — 2 sensors')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    print(sensor_A.shape) # Should be (200,)

if __name__ == '__main__':
    main()
```

```python
# Scatter plot: sensor reading (y-axis) vs timestamp in seconds (x-axis).
# Color Sensor A points blue and Sensor B points orange.
# Add axis labels with units, a title, and a legend.

import matplotlib.pyplot as plt
import numpy as np

n = 200
seed = 1234 # replace with the last 4 digits of your Drexel ID
rng = np.random.default_rng(seed)

timestamps = np.sort(rng.uniform(0.0, 10.0, n)).astype(np.float64)
sensor_a = rng.normal(loc=25.0, scale=3.0, size=n).astype(np.float64)
sensor_b = rng.normal(loc=27.0, scale=4.5, size=n).astype(np.float64)

print(timestamps.shape, sensor_a.shape, sensor_b.shape) # Should be (200,) for all three arrays

plt.figure(figsize=(9,5))
plt.scatter(timestamps, sensor_a, c='C0', marker='o', s=36, alpha=0.75, label='Sensor A (μ=25, σ=3)')
plt.scatter(timestamps, sensor_b, c='C1', marker='s', s=36, alpha=0.75, label='Sensor B (μ=27, σ=4.5)')
plt.xlabel('Time (s)')
plt.ylabel('Temperature (°C)')
plt.title('Simulated temperature readings (scatter)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
# Overlaid histogram of Sensor A and Sensor B temperature distributions.
# Use 30 bins, alpha=0.5 for transparency so both distributions are visible.
# Add vertical dashed lines at each sensor's mean.
# Include a legend labeling each sensor.

import matplotlib.pyplot as plt
import numpy as np

try:
    print("sensor_a shape:", sensor_a.shape)
    print("sensor_a dtype:", sensor_a.dtype)
    print("sensor_b shape:", sensor_b.shape)
    print("sensor_b dtype:", sensor_b.dtype)
except NameError:
    raise NameError("sensor_a and/or sensor_b not found. Run the data-generation cell first.")

print("sensor_a mean/std: {:.3f} / {:.3f}".format(np.mean(sensor_a), np.std(sensor_a)))
print("sensor_b mean/std: {:.3f} / {:.3f}".format(np.mean(sensor_b), np.std(sensor_b)))

all_data = np.concatenate([sensor_a, sensor_b])
bins = np.histogram_bin_edges(all_data, bins='auto')

plt.figure(figsize=(8,5))
plt.hist(sensor_a, bins=bins, alpha=0.6, color='C0', label='Sensor A', edgecolor='black')
plt.hist(sensor_b, bins=bins, alpha=0.6, color='C1', label='Sensor B', edgecolor='black')
plt.xlabel('Temperature (°C)')
plt.ylabel('Count')
plt.title('Histogram of Sensor A and Sensor B readings')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
